```python
class Deporte:
    def __init__(self, nombre, numJugadores, equipo):
        self.nombre = nombre
        self.numJugadores = numJugadores
        self.equipo = equipo

    def leerDatos(self):
        self.nombre = input("Ingrese el nombre del deporte: ")
        self.numJugadores = input("Ingrese el número de jugadores: ")
        self.equipo = input("Ingrese el equipo: ")
        return self.nombre, self.numJugadores, self.equipo

    def mostrarDatos(self):
        print("Nombre: ", self.nombre)
        print("Número de Jugadores: ", self.numJugadores)
        print("Equipo: ", self.equipo)
        return self.nombre, self.numJugadores, self.equipo

# fin de la clase Deporte

print("deporte numero 1")
deporte1 = Deporte("", "", "") # creando el objeto deporte1
deporte1.leerDatos()
deporte1.mostrarDatos()

print("*_*_*_*_*_*_+_*_*_*_*_*_*")

print("deporte numero 2")
deporte2 = Deporte("", "", "") # creando el objeto deporte2
deporte2.leerDatos()
deporte2.mostrarDatos()

print("software desarrollado por Hugo Fabian NL = 15 )")
```
